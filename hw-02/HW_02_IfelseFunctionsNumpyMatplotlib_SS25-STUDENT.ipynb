## 3. Functions, if/else, and Manufacturing! (10 Points)

Different products require different surface finish. For example, consumer products are often expected to be smooth and shiny, while non-critical components of boilers can be rough. Certain industries have stricter requirements on surface finish and tolerance such as the aerospace industry. However, tighter tolerances and smoother surfaces come at the cost of increased production expenses and longer production times, see [here](https://web.mae.ufl.edu/designlab/Lab%20Assignments/EML2322L-Tolerances.pdf). 

Different production processes have limitations on the surface finish (measured in micrometers $\mu m$ which is one millionth of a meter, or microns). For example, [grinding](https://en.wikipedia.org/wiki/Grinding_(abrasive_cutting)) can produce finer surfaces than [turning](https://en.wikipedia.org/wiki/Turning) or [milling](https://en.wikipedia.org/wiki/Milling_(machining)), see Fig. 15.8 [here](https://web.mae.ufl.edu/designlab/Lab%20Assignments/EML2322L-Tolerances.pdf).

Rough ranges for surface finished produced by turning, milling, and grinding are:
* Grinding: 0.025 $\mu m$ -- 3.2 $\mu m$.
* Turning: 0.40 $\mu m$ -- 3.2 $\mu m$.
* Milling: 0.40 $\mu m$ -- 25 $\mu m$.


### 3.1 Selecting the right process for the job  (3 points)

Write a function that takes in the desired surface finish (in microns) and a manufacturing process as a string ('turning', 'milling', or 'grinding') and returns the Boolean "True" if the process can produce the required surface finish, and definition and "False" if it cannot. 

Write your function in the cell below. Demonstrate that your function works using the test data given (`surface_finish_test_data`).

__hint__: You can test if a string is in a list using ```in```. For example, ``` 'cat' in ['dog', 'dolphin', 'cat']``` will evaluate to ```True```.

````python
#----------------Run this cell before proceeding, but do no edit it.
# test surface finish (in microns)
surface_finish_test_data = [11.56,  0.78,  0.11,  0.96, 12.33,  0.84, 13.42,  0.91,  0.13, 10.45,  0.09,  0.10 ]  # in microns
````

````python
# Write your Answer to 3.1 here:
def finish(surface_finish, process):
    surface_finish_ranges = {
        "grinding": (0.025, 3.2),
        "turning": (0.40, 3.2),
        "milling": (0.40, 25) }
    
    if process in surface_finish_ranges:
        min_sf, max_sf = surface_finish_ranges[process]
        return min_sf <= surface_finish <= max_sf
    else:
        return False  

print(finish(surface_finish_test_data[0],"grinding"))
print(finish(surface_finish_test_data[0],"turning"))
print(finish(surface_finish_test_data[0],"milling"))



````

````python
#--------------Run this cell, but do not edit it----------------------------
# function to estimate the amount of time it would take to finish to a required tolerance using a specific machining process

def estimate_time(surface_finish, process):
    if process == 'turning':
        process_time = 8.030 * np.exp(1.031 * surface_finish)
    elif process == 'milling':
        process_time = 11.926 * np.exp(0.472 * surface_finish)
    elif process == 'grinding':
        process_time = 21.669 * np.exp(17.150 * surface_finish)
    else:
        print('incorrect input process')
        return
    
    # return the estiamted time rounded to 3 decimal places
    return np.round(process_time,3)
#-------------------Write your answer to 3.3 in the following cell---------
````

````python
# write you answer to 3.3 here:
def recommend_process(surface_finish):
    processes = ["turning", "milling", "grinding"]
    viable_processes = []
    for p in processes:
        if finish(surface_finish, p):
            viable_processes.append(p)
    if not viable_processes:
        return None  

    best_process = None
    best_time = float("inf")

    for p in viable_processes:
        time = estimate_time(surface_finish, p)
        if time < best_time:
            best_time = time
            best_process = p
    return best_process


process_counts = {"turning": 0, "milling": 0, "grinding": 0}

for tolerance in tolerance_list:
    recommended = recommend_process(tolerance)
    if recommended:
        process_counts[recommended] += 1
    
    
print("Recommended Process Tally")
print("Turning    Milling    Grinding")
print(f"  {process_counts['turning']}         {process_counts['milling']}          {process_counts['grinding']}")
````

### 4.2 Using `numpy` arrays and functions for calculations (6 points)

The temperature anomalies in different regions vary significantly over time. To better compare trends between regions, we can examine the percent change in temperature anomalies rather than absolute values.

We can calculate the percent change as follows:

$\% \ difference = \frac{(final-initial)}{initial} \times 100$

&#9989;&nbsp; In the cell below, write a function that takes in an array of temperature anomalies and calculates the percent change between each value. Demonstrate that your code works using one of the temperature anomaly arrays from the dataset.

````python
# put your code here
def calculate_percent_change(temp_anomalies):
    percent_changes = []
    
    for i in range(1, len(temp_anomalies)):
        initial = temp_anomalies[i - 1]
        final = temp_anomalies[i]
        if initial == 0:
            percent_change = float('nan')  
        else:
            percent_change = ((final - initial) / initial) * 100
        
        percent_changes.append(percent_change)
    
    return np.array(percent_changes)

global_temp_percent_change = calculate_percent_change(global_temp)

print("Years\tGlobal Temp % Change")

for i in range(len(global_temp_percent_change)):
    year = years[i + 1]  
    percent_change = global_temp_percent_change[i]
    print(f"{year}\t{percent_change:.2f}%")
````
